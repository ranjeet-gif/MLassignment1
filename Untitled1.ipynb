```python
import pandas as pd
```

```python
df = pd.read_csv('/content/crocodile_dataset (1).csv')
display(df.head())
```

```python
display(df.info())
display(df.describe())
```

## Check for missing values



```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate the number of missing values for each column
missing_values_count = df.isnull().sum()

# Display the count of missing values for each column
print("Missing values count per column:")
display(missing_values_count)

# Calculate the percentage of missing values for each column
missing_values_percentage = (df.isnull().sum() / len(df)) * 100

# Display the percentage of missing values for each column
print("\nMissing values percentage per column:")
display(missing_values_percentage)

# Create a heatmap to visualize the distribution of missing values
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Heatmap of Missing Values')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

categorical_cols = ['Common Name', 'Scientific Name', 'Genus', 'Age Class', 'Sex', 'Country/Region', 'Habitat Type', 'Conservation Status']

for col in categorical_cols:
    plt.figure(figsize=(10, len(df[col].unique()) * 0.5))
    sns.countplot(data=df, y=col, order=df[col].value_counts().index, palette='viridis')
    plt.title(f'Distribution of {col}')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```
